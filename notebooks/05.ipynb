### Q1 Quarantine versus delete
> Why keep bad orders in a separate file rather than deleting them? Who would check that file at a real company?
We keep bad orders in quarantine instead of deleting them for auditing, debugging and potential data‑recovery.
If we delete bad data directly, we lose evidence of data‑quality problems and cannot investigate root causes.
In a real company, data engineers or data quality analysts would review the quarantine file, identify upstream source‑system bugs, and fix or recover valid records.



### 2 Hash versus mask

> 
> email_hash lets analysts count unique customers; email_masked is for display. Why would hashing without a salt be weak for emails?

email_hash is for analytics to count distinct customers without exposing real emails.
email_masked is for human‑readable display.
Hashing without a salt is weak because attackers can use pre‑computed rainbow tables to reverse‑lookup many common email hashes and recover original email addresses.

> 
> Which PDPA 2010 principle does this follow, and why is age_group safer than birth_year?
This follows the PDPA 2010 principle of data minimisation: only collect and retain personal data necessary for business purpose.
age‑group is safer than birth‑year because it reduces precision of personal information. It prevents exact age calculation and makes individual harder to re‑identify.
