```python
import sys
sys.path.append("..")
import pandas as pd

books = pd.read_csv("../data/lake/bronze/books.csv")
customers = pd.read_csv("../data/lake/bronze/customers.csv")
orders = pd.read_csv("../data/lake/bronze/orders.csv")
reviews = pd.read_csv("../data/lake/bronze/reviews.csv")

print("books shape", books.shape)
print("customers shape", customers.shape)
print("orders shape", orders.shape)
print("reviews shape", reviews.shape)
```

Answers
1. What does one row mean in each file?
- books: one individual book
- customers: one customer
- orders: one order record
- reviews: one customer review for a book

2. Which columns link the files together?
`customer_id` links customers, orders and reviews.
`book_id` links books, orders and reviews.

3. Which columns hold personal data?
Name, email, phone number and MyKad number in customers.csv are personal data.

4. Which source would you collect once, and which every day?
Collect once: books catalogue, customer base data.
Collect every day: exchange rate, new orders and new reviews.
